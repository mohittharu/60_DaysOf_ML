Sigmoid
-----------------
Sigmoid converts a value into a number between:

0 and 1

Formula:

σ(x) = 1 / (1 + e⁻ˣ)

Don't worry about memorizing the formula right now.

Think:

Large negative → close to 0

0 → 0.5

Large positive → close to 1

Examples:

Sigmoid(-10) ≈ 0
Sigmoid(0)   = 0.5
Sigmoid(10)  ≈ 1

```python
# simple program to demonstrate changing weights in a neural network

import math 

def sigmoid(x):
    return 1 / (1 + math.exp(-x))


print(sigmoid(-5))  # Output: 0.0066928509242848554

print(sigmoid(0))   # Output: 0.5

print(sigmoid(5))   # Output: 0.9933071490757153
```

# When Do We Use Sigmoid?

Most importantly:

Binary classification

Example:

Is this email spam?

0 → Not Spam
1 → Spam

The model might produce:

0.92

We can interpret that approximately as:

92% probability of the positive class.

So a typical network can be:

Input
 ↓
Hidden Layers + ReLU
 ↓
Output + Sigmoid
 ↓
0 to 1

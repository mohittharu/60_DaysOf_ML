ReLU
----------------

ReLU = Rectified Linear Unit

Formula:

ReLU(x) = max(0, x)

Meaning:

If x < 0 → 0
If x > 0 → x

Examples:

ReLU(-5) = 0
ReLU(-2) = 0
ReLU(0)  = 0
ReLU(3)  = 3
ReLU(10) = 10
Visual idea
Negative values → 0
Positive values → unchanged

```python
# simple python code 
def relu(x):
    return max(0, x)

print(relu(-5))  # Output: 0
print(relu(3))   # Output: 3
```

Why is ReLU So Popular?

ReLU is:

Simple
Fast
Computationally efficient
Usually good for hidden layers
Helps reduce some problems associated with sigmoid/tanh

That's why you'll see:

Dense → ReLU

very often in Deep Learning models.

# Softmax 

Softmax is extremely important for multi-class classification.

Suppose we have:

Cat
Dog
Horse

The network might produce raw scores:

Cat   = 2.0
Dog   = 1.0
Horse = 0.1

Softmax converts them into probabilities.

Something like:

Cat   = 0.66
Dog   = 0.24
Horse = 0.10

And importantly:

0.66 + 0.24 + 0.10 = 1.00

So:

Softmax converts multiple scores into probabilities whose total is 1.

## Sigmoid vs Softmax

This is something you should remember.

Problem	Output Activation
Binary classification	Sigmoid
Multi-class classification	Softmax
Hidden layers	ReLU usually
Some RNN/LSTM components	Tanh/Sigmoid

Example:

Binary
Dog vs Not Dog
       ↓
    Sigmoid
       ↓
     0.87
Multi-class
Cat / Dog / Horse
       ↓
    Softmax
       ↓
Cat:   0.10
Dog:   0.80
Horse: 0.10

```python
# Build activation functions using numpy and matplotlib


import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)

relu = np.maximum(0, x)

sigmoid = 1 / (1 + np.exp(-x))

tanh = np.tanh(x)

plt.figure(figsize=(8, 5))

plt.plot(x, relu, label='ReLU', color='blue')
plt.plot(x, sigmoid, label='Sigmoid', color='green')
plt.plot(x, tanh, label='Tanh', color='red')

plt.legend()
plt.xlabel("Input")
plt.ylabel("Output")
plt.title("Activation Functions")

plt.grid()
plt.show()
```

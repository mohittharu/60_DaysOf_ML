Why Do We Need Activation Functions?

This is extremely important.

Suppose a neuron calculates:

z = x₁w₁ + x₂w₂ + b

If we simply pass z to the next layer, the neural network becomes basically a sequence of linear operations.

That limits what it can learn.

Activation functions introduce non-linearity.

Simple meaning:

An activation function helps a neural network learn complicated patterns.

## Tanh

Tanh is similar to sigmoid but produces values between:

-1 and +1

Examples:

tanh(-∞) → -1
tanh(0)  → 0
tanh(+∞) → +1

```python
# simple program to demonstrate changing weights in a neural network

import math 
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))  

print(tanh(-5))  # Output: -0.9999092042625951

print(tanh(0))   # Output: 0.0

print(tanh(3))   # Output: 0.9950547536867305
```

## Where is Tanh used?

Historically/common use:

RNNs
LSTMs
Some neural network architectures

For many ordinary hidden layers today, ReLU-family activations are more common.

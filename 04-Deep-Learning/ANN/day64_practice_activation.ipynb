```python
# Predict the output before running:
def relu(x):
    return max(0, x)

numbers = [-10, -5, -1, 0, 2, 5, 10]

for n in numbers:
    print(n, "→", relu(n))
```

```python
# sigmoid 

import math 

def sigmoid(x):
    return 1 / (1 + math.exp(-x))

numbers = [-10, -5, -1, 0, 2, 5, 10]

for n in numbers:
    print(n, "→", sigmoid(n))
```

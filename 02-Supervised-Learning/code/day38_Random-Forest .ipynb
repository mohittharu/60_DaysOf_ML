```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Load dataset 
data = load_breast_cancer()
x= pd.DataFrame(data.data, columns = data.feature_names)
y= pd.Series(data.target)
```

```python
# Train-test split
x_train, x_test , y_train, y_test = train_test_split(
    x,y, test_size=0.2, random_state=4, stratify =y
)
```

```python
# Decision tree without pruning (max_depth= none)
deep_tree = DecisionTreeClassifier(random_state=42)
deep_tree.fit(x_train, y_train )
```

```python
from sklearn.ensemble import RandomForestClassifier

# Random Forest (ensemble of 100 trees)
rf_model = RandomForestClassifier(
    n_estimators=100, 
    max_depth=None,   # each tree can grow deep, but averaging reduces overfitting
    random_state=42
)
rf_model.fit(x_train, y_train)

# Evaluate
train_acc_rf = accuracy_score(y_train, rf_model.predict(x_train))
test_acc_rf = accuracy_score(y_test, rf_model.predict(x_test))

print("\nRandom Forest:")
print("Training Accuracy:", train_acc_rf)
print("Test Accuracy:", test_acc_rf)
```

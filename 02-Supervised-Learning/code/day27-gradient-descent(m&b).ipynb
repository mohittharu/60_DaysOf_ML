```python
from sklearn.datasets import make_regression

import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt 
```

```python
%matplotlib inline
from matplotlib.animation import FuncAnimation 
import matplotlib.animation as animation
```

```python
# Generate regression data
X, y = make_regression(
    n_samples=100,
    n_features=1,
    n_informative=1,
    n_targets=1,
    noise=20,
    random_state=13
)
```

```python
# Scatter plot
fig, ax = plt.subplots()
ax.scatter(X, y, alpha=0.7)
ax.set_xlabel("Feature X")
ax.set_ylabel("Target y")
ax.set_title("Synthetic Regression Dataset")

plt.show()
```

```python

# 3. Gradient Descent
# ===============================
b = -520
m = 600
lr = 0.001
epochs = 30

all_b = []
all_m = []
all_cost = []

n = len(X)

for i in range(epochs):
    db = 0
    dm = 0
    cost = 0

    for j in range(n):
        y_pred = m * X[j] + b
        error = y[j] - y_pred

        db += -2 * error
        dm += -2 * error * X[j]
        cost += error ** 2

    db /= n
    dm /= n
    cost /= n

    b -= lr * db
    m -= lr * dm

    all_b.append(b)
    all_m.append(m)
    all_cost.append(cost)
```

```python
# ===============================
# 4. Animation
# ===============================
fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(X, y, alpha=0.7)

x_i = np.arange(X.min() - 1, X.max() + 1, 0.1)

line, = ax.plot(x_i, all_m[0] * x_i + all_b[0], 'r-', linewidth=2)

ax.set_ylim(y.min() - 50, y.max() + 50)
ax.set_title("Linear Regression using Gradient Descent")

def update(i):
    line.set_ydata(all_m[i] * x_i + all_b[i])
    ax.set_xlabel(f"Epoch {i+1}")
    return line,

anim = FuncAnimation(
    fig,
    update,
    frames=epochs,
    interval=500,
    repeat=True
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# ===============================
# Cost Function Animation
# ===============================

epochs = len(all_cost)
epoch_nums = list(range(1, epochs + 1))

fig, ax = plt.subplots(figsize=(9, 5))
ax.set_xlim(1, epochs)
ax.set_ylim(0, max(all_cost) * 1.1)
ax.set_title("Cost Function vs Epochs")
ax.set_ylabel("Cost")

line, = ax.plot([], [], lw=2)

xdata, ydata = [], []

def animate(i):
    xdata.append(epoch_nums[i])
    ydata.append(all_cost[i])

    line.set_data(xdata, ydata)
    ax.set_xlabel(f"Epoch {i+1}")
    return line,

anim = animation.FuncAnimation(
    fig,
    animate,
    frames=epochs,
    interval=500,
    repeat=False
)

plt.show()

# ===============================
# Save animation as GIF
# ===============================
writergif = animation.PillowWriter(fps=2)
anim.save("animation5.gif", writer=writergif)
```

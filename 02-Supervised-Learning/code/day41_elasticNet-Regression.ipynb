# ElasticNet Regression (Regularization)

## Elastic Net Regression is a regularized linear regression technique 
## that combines the penalties of Lasso Regression and Ridge Regression.
## It is mainly used to improve prediction accuracy and prevent overfitting,
## especially when dealing with many correlated features.

```python
# code 
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression

# generate dataset
X, y = make_regression(n_samples=100, n_features=20, noise=0.1)

# train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# create model
model = ElasticNet(alpha=1.0, l1_ratio=0.5)

# train
model.fit(X_train, y_train)

# coefficients
print(model.coef_)
```
